```python
# SQL Library and Pandas
import sqlite3
import pandas as pd

# Connect to the database
conn = sqlite3.connect("data.sqlite")
```

```python
pd.read_sql("""SELECT * FROM employees""", conn)
```

```python
pd.read_sql("""
SELECT employeeNumber, lastName
FROM employees;
""", conn).head(10)
```

```python
pd.read_sql("""
SELECT lastName, employeeNumber
FROM employees;
""", conn).head(10)
```

```python
pd.read_sql("""
SELECT lastName, employeeNumber AS ID
FROM employees;
""", conn).head(10)
```

```python
pd.read_sql("""
SELECT *,
    CASE
        WHEN jobTitle = 'President'
          OR jobTitle = 'VP Sales'
          OR jobTitle = 'VP Marketing'
        THEN 'Executive'
        ELSE 'Not Executive'
    END AS role
FROM employees;
""", conn)
```

```python
pd.read_sql("""
SELECT LENGTH(lastName) AS name_length
FROM employees;
""", conn)
```

```python
pd.read_sql("""
SELECT SUBSTR(jobTitle, 1, 2) AS short_title
FROM employees;
""", conn)
```

```python
pd.read_sql("""
SELECT ROUND(priceEach * quantityOrdered) AS total_price
FROM orderDetails;
""", conn)
```

```python
pd.read_sql("""
SELECT ROUND(priceEach * quantityOrdered) AS total_price
FROM orderdetails;
""", conn).sum()
```

```python
pd.read_sql("""
SELECT
    orderDate,
    STRFTIME('%d', orderDate) AS day,
    STRFTIME('%m', orderDate) AS month,
    STRFTIME('%Y', orderDate) AS year
FROM orders;
""", conn)
```
